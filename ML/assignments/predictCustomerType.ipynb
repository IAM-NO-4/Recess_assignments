Predicting whether customer is a Wholesaler or Retailer

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
```

Data gathering and cleaning

```python
df = pd.read_csv("DSA_ecommerce_bigdata.csv")

df.isnull().sum()

df.duplicated().sum()

df.describe()
```

Encoding 

```python
encoded_df = pd.get_dummies(df,columns=['Region','ProductCategory'])
encoded_df
```

splitting data

```python
y = encoded_df['CustomerType']
y
X = encoded_df.drop(columns=['CustomerType','TransactionID','Date'])
#X


X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2, random_state=20)
```

Model selection and training

```python
log_reg = LogisticRegression(random_state=20,max_iter=5000)
dec_tree = DecisionTreeClassifier(random_state=20)
rand_forest = RandomForestClassifier(random_state=20)

models = {
    "Logistics Model":log_reg,
    "Decision Tree":dec_tree,
    "Random Forest":rand_forest
}


for name,model in models.items():
    model.fit(X_train,y_train)
    accuracy = model.score(X_test,y_test)
    y_pred = model.predict(X_test)
    
    print(f"model: {name}\nAccuracy Score: {accuracy}")
    print("---------Classification Report ------------")
    print(classification_report(y_test,y_pred))
    
```
